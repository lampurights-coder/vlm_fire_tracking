# 🧠 **LoRA (Low-Rank Adaptation) - The Magic Behind Efficient Fine-Tuning**

## 🎯 **What is LoRA?**
**LoRA = Low-Rank Adaptation** - A parameter-efficient fine-tuning technique that **freezes the original weights** and injects **trainable rank decomposition matrices** into transformer layers.

```
┌─────────────────────────────────────────────────────────────┐
│ 🎭 BEFORE LoRA                                              │
│                                                             │
│   W (Original Weights) - Frozen ❄️                         │
│   │                                                        │
│   ▼                                                        │
│   [LLM/Vision Model] - 8B Parameters = Heavy 🏋️           │
└─────────────────────────────────────────────────────────────┘

┌─────────────────────────────────────────────────────────────┐
│ 🚀 WITH LoRA                                                │
│                                                             │
│   W (Original) + ΔW (LoRA) = W' (Updated)                  │
│   Where ΔW = BA^T                                          │
│   │         │  B: d × r                                    │
│   │         │  A: r × k                                    │
│   │         └─ r ≪ min(d,k) (Low Rank!)                   │
│   ▼                                                        │
│   [LLM/Vision Model] + Tiny Adapter = Smart & Light 🎈     │
└─────────────────────────────────────────────────────────────┘
```

## 🧩 **The Mathematics Behind LoRA**

### 📐 **Original Forward Pass:**
```
h = W₀x + b
```
Where:
- **W₀**: Original weight matrix (d × k) - **Frozen!** ❄️
- **x**: Input
- **h**: Output

### 🎯 **With LoRA:**
```
h = W₀x + (B × A)x + b
```
Where:
- **B**: Down projection matrix (d × r) - **Trainable!** 🔥
- **A**: Up projection matrix (r × k) - **Trainable!** 🔥
- **r**: Rank (typically 4, 8, 16, 32) - **TINY!** 🐜

### 🎪 **Visual Representation:**
```
Original: [d × k] = 4096 × 4096 = 16.7M parameters ❗
LoRA:     [d × r] + [r × k] = (4096×16) + (16×4096) = 131K parameters 🎉
```
**Parameter Reduction:** **99.2% Less Parameters!** 🤯

## 🔧 **Your Configuration Explained**

```python
# 🎛️ LoRA Configuration Parameters
r = 16,           # 🎯 Rank - The "intelligence capacity" of adapter
                  # Higher r = More expressive = More parameters = Potentially overfit 🎭
                  # Lower r = Less expressive = Fewer parameters = Less overfit 🛡️

lora_alpha = 16,  # ⚖️ Scaling factor - How much LoRA affects the original weights
                  # Rule: alpha == r (empirically best) ⚖️
                  # Final scaling = alpha/r = 16/16 = 1.0 ✅

lora_dropout = 0, # 🚫 No dropout - Full gradient flow (can add regularization if overfitting)

bias = "none",    # ⚡ Don't train bias terms - Focus only on weight matrices

random_state = 3407, # 🎲 Seed for reproducibility - Same results every time! 🔒

use_rslora = False, # 🔄 Rank-Stabilized LoRA - Advanced technique for better stability
                    # False = Standard LoRA (usually works great!)

loftq_config = None, # 🎯 LoftQ - Quantization-aware LoRA initialization
                     # None = Standard initialization (works with 4-bit base model!)
```

### 🎭 **Target Modules Breakdown:**
```
┌─────────────────────────────────────────┐
│ 🎯 ATTENTION LAYER ADAPTATION           │
├─────────────────────────────────────────┤
│ Original: Q, K, V, O matrices          │
│ LoRA: Adds ΔQ, ΔK, ΔV, ΔO              │
│                                        │
│ Q' = Q + B_q × A_q                     │
│ K' = K + B_k × A_k                     │
│ V' = V + B_v × A_v                     │
│ O' = O + B_o × A_o                     │
└─────────────────────────────────────────┘

┌─────────────────────────────────────────┐
│ 🧠 MLP/FFN LAYER ADAPTATION             │
├─────────────────────────────────────────┤
│ Original: W₁, W₂ matrices              │
│ LoRA: Adds ΔW₁, ΔW₂                    │
│                                        │
│ W₁' = W₁ + B₁ × A₁                     │
│ W₂' = W₂ + B₂ × A₂                     │
└─────────────────────────────────────────┘
```

## 🔬 **Expert Insights**

### 🎯 **Rank Choice Strategy:**
- **r=4-8**: Good for simple adaptations 🐥
- **r=16-32**: Balanced for most tasks 🐦
- **r=64+**: Complex adaptations, risk overfitting 🦅

### ⚡ **Alpha Scaling:**
```
ΔW_scaled = (alpha/r) × BA^T
```
- **alpha = r**: Standard scaling (your choice!) ✅
- **alpha < r**: Less influence from LoRA 🔽
- **alpha > r**: More influence from LoRA 🔼

### 🔄 **Gradient Flow:**
```
┌─────────────────────────────────────────┐
│ 🎯 GRADIENT PROPAGATION                 │
├─────────────────────────────────────────┤
│ Only B and A get gradients!            │
│ W₀ stays frozen ❄️                     │
│                                        │
│ ∂Loss/∂B = (∂Loss/∂h) × A^T × x^T     │
│ ∂Loss/∂A = B^T × (∂Loss/∂h) × x^T     │
└─────────────────────────────────────────┘
```

### **Step 3: 🎯 YOLO → Qwen-VL Format Conversion**

#### **🔢 YOLO Format (Normalized Coordinates):**
```
class_id center_x center_y width height
0 0.52 0.38 0.12 0.24
```
Where values are **normalized** (0-1) relative to image dimensions.

#### **🎯 Conversion Process:**
```python
# 🧮 Math behind the conversion
x1 = int((cx - w/2) * 1000)  # Left
y1 = int((cy - h/2) * 1000)  # Top  
x2 = int((cx + w/2) * 1000)  # Right
y2 = int((cy + h/2) * 1000)  # Bottom

# 🔒 Clamp to [0, 999] for Qwen-VL's 1000×1000 coordinate system
x1 = max(0, min(999, x1))
y1 = max(0, min(999, y1))
x2 = max(0, min(999, x2))
y2 = max(0, min(999, y2))
```

**Example Conversion:**
```
YOLO: fire 0.52 0.38 0.12 0.24 (normalized 640×640)
→ Qwen-VL: fire<loc520><loc380><loc640><loc620> (scaled to 1000×1000)
```

### **🧮 Mathematical Foundations**

#### **📏 Intersection over Union (IoU)**
```python
def iou(box1, box2):
    """
    🧮 IoU = Intersection Area / Union Area
    
    box1: [x1, y1, x2, y2]
    box2: [x1, y1, x2, y2]
    
    Returns: 0.0 to 1.0 (1.0 = perfect overlap)
    """
    xi1 = max(box1[0], box2[0])  # Intersection left
    yi1 = max(box1[1], box2[1])  # Intersection top
    xi2 = min(box1[2], box2[2])  # Intersection right
    yi2 = min(box1[3], box2[3])  # Intersection bottom
    
    inter_area = max(0, xi2 - xi1) * max(0, yi2 - yi1)
    union_area = (box1_area + box2_area - inter_area)
    
    return inter_area / union_area if union_area > 0 else 0
```

```python
    pil_img = Image.fromarray(cv2.cvtColor(cv_img, cv2.COLOR_BGR2RGB))
    
    # Step 2: Model Inference
    output_ids = model.generate(**inputs, max_new_tokens=128, temperature=0.7)
    generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=False)
    
    # Step 3: Parse Bounding Boxes
    matches = bbox_pattern.findall(generated_text)
    dets = []
    for match in matches:
        label = match[0]
        x1, y1, x2, y2 = map(int, match[1:])
        # Scale from 0-1000 to image size
        x1 = int(x1 / 1000 * width)
        y1 = int(y1 / 1000 * height)
        x2 = int(x2 / 1000 * width)
        y2 = int(y2 / 1000 * height)
        bbox = np.array([x1, y1, x2, y2])
        dets.append((label, bbox))
    
    # Step 4: Smart Label Refinement
    fires = [bbox for label, bbox in dets if label == 'fire']
    others = [bbox for label, bbox in dets if label != 'fire']
    
    # 🔄 IoU-based relabeling (fix model mistakes)
    for other in others:
        for prev in previous_fires:
            if iou(other, prev) > iou_threshold:
                fires.append(other)  # Relabel as fire
                break
    
    # Step 5: Prepare for Tracking
    dets_np = np.array([[*bbox, 0.8, 0] for bbox in fires]) if fires else np.zeros((0, 6))
    # Format: [x1, y1, x2, y2, confidence, class_id]
    
    # Step 6: Update Tracker
    tracks = tracker.update(dets_np, cv_img)
    
    return cv_img, tracks
```

````python
def save_frames_from_video(video_path, num_frames_to_save, output_folder="frames_output"):
    # Create output folder if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)

    # Open the video
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print("Error: Could not open video.")
        return

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(f"Total frames in video: {total_frames}")

    saved_count = 0
    frame_index = 0

    while saved_count < num_frames_to_save:
        ret, frame = cap.read()

        # If video ended early
        if not ret:
            print("Reached end of video.")
            break

        # Save the frame
        frame_path = os.path.join(output_folder, f"frame_{frame_index}.jpg")
        cv2.imwrite(frame_path, frame)

        saved_count += 1
        frame_index += 1

    cap.release()
    print(f"Saved {saved_count} frames to {output_folder}")

# -----------------------------
# Example usage
# -----------------------------
video_path = "/home/ai/Desktop/PADDLE/fire.webm"
num_frames_to_save = 500

save_frames_from_video(video_path, num_frames_to_save)
````

````python

# ========================= CONFIG =========================
input_video_path = "output.mp4" # Change to your video path
output_video_path = "./fire_detection_output.mp4"
target_fps = 15 # Process and output at 15 FPS
instruction = "detect fire bbox" # Your prompt
# Model & tokenizer (already loaded in your environment)
# Assuming you have:
# model = FastVisionModel.for_inference(model)
# tokenizer
# Regex to parse multiple: e.g., fire<loc479><loc318><loc530><loc373> other<loc...> <|im_end|>
bbox_pattern = re.compile(r'(\w+)<loc(\d+)><loc(\d+)><loc(\d+)><loc(\d+)>')
iou_threshold = 0.5  # High IoU threshold for relabeling
# ========================= FUNCTIONS =========================
def iou(box1, box2):
    """Compute IoU between two bounding boxes [x1, y1, x2, y2]"""
    xi1 = max(box1[0], box2[0])
    yi1 = max(box1[1], box2[1])
    xi2 = min(box1[2], box2[2])
    yi2 = min(box1[3], box2[3])
    inter_area = max(0, xi2 - xi1) * max(0, yi2 - yi1)
    box1_area = (box1[2] - box1[0]) * (box1[3] - box1[1])
    box2_area = (box2[2] - box2[0]) * (box2[3] - box2[1])
    union_area = box1_area + box2_area - inter_area
    return inter_area / union_area if union_area > 0 else 0

def process_single_frame(cv_img, tracker, previous_fires):
    if cv_img is None:
        return None, previous_fires
    height, width = cv_img.shape[:2]
    pil_img = Image.fromarray(cv2.cvtColor(cv_img, cv2.COLOR_BGR2RGB))
    # Prepare input
    messages = [
        {"role": "user", "content": [
            {"type": "image"},
            {"type": "text", "text": instruction}
        ]}
    ]
    input_text = tokenizer.apply_chat_template(messages, add_generation_prompt=True)
    inputs = tokenizer(pil_img, input_text, add_special_tokens=False, return_tensors="pt").to("cuda")
    # Generate
    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=128,
            temperature=0.7,
            do_sample=True,
            use_cache=True
        )
    generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=False)
    # Extract all bboxes from generated text
    matches = bbox_pattern.findall(generated_text)
    dets = []
    for match in matches:
        label = match[0]
        x1, y1, x2, y2 = map(int, match[1:])
        # Scale from 0-1000 to image size
        x1 = int(x1 / 1000 * width)
        y1 = int(y1 / 1000 * height)
        x2 = int(x2 / 1000 * width)
        y2 = int(y2 / 1000 * height)
        bbox = np.array([x1, y1, x2, y2])
        dets.append((label, bbox))
    if not matches:
        print(f"No valid bboxes found in: {generated_text}")
    # Filter fires and relabel others with high IoU to previous fires
    fires = [bbox for label, bbox in dets if label == 'fire']
    others = [bbox for label, bbox in dets if label != 'fire']
    for other in others:
        for prev in previous_fires:
            if iou(other, prev) > iou_threshold:
                fires.append(other)
                break
    # Prepare detections for tracker: [x1, y1, x2, y2, conf, cls]
    dets_np = np.array([[*bbox, 0.8, 0] for bbox in fires]) if fires else np.zeros((0, 6))
    # Update tracker
    tracks = tracker.update(dets_np, cv_img)
    # Draw tracks on cv_img
    for track in tracks:
        x1, y1, x2, y2, track_id = map(int, track[:5])
        cv2.rectangle(cv_img, (x1, y1), (x2, y2), (0, 0, 255), 5)  # Red box
        text = f"fire ID:{track_id}"
        cv2.rectangle(cv_img, (x1, y1 - 30), (x1 + len(text) * 15, y1), (0, 0, 255), -1)  # Text background
        cv2.putText(cv_img, text, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.75, (255, 255, 255), 2)
    # Update previous_fires with current tracked fire bboxes
    previous_fires = [track[:4] for track in tracks]
    return cv_img, previous_fires
# ========================= MAIN PROCESSING =========================
# Open input video
cap = cv2.VideoCapture(input_video_path)
if not cap.isOpened():
    raise ValueError(f"Cannot open video: {input_video_path}")
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
# Get accurate duration using ffprobe
try:
    cmd = ['ffprobe', '-v', 'error', '-show_entries', 'format=duration', '-of', 'default=noprint_wrappers=1:nokey=1', input_video_path]
    duration = float(subprocess.check_output(cmd).decode('utf-8').strip())
    input_fps = total_frames / duration if duration > 0 else cap.get(cv2.CAP_PROP_FPS)
except Exception as e:
    print(f"Failed to get duration with ffprobe: {e}. Falling back to OpenCV FPS.")
    input_fps = cap.get(cv2.CAP_PROP_FPS)
if input_fps <= 0:
    input_fps = 30.0  # Fallback to a reasonable default if still invalid
frame_interval = max(1, int(round(input_fps / target_fps)))
estimated_processed_frames = total_frames // frame_interval
print(f"Input video FPS: {input_fps}, Target FPS: {target_fps}, Frame interval: {frame_interval}")
print(f"Estimated frames to process: {estimated_processed_frames}")
# Initialize tracker
device = 'cuda' if torch.cuda.is_available() else 'cpu'
tracker = BotSort(
    reid_weights="",
    device=device,
    fp16=False,
    half=True,
    track_high_thresh=0.5,
    track_low_thresh=0.1,
    new_track_thresh=0.6,
    track_buffer=30,
    match_thresh=0.8,
    proximity_thresh=0.5,
    appearance_thresh=0.25,
    with_reid=False,
    frame_rate=target_fps
)
# Initialize video writer
video_writer = cv2.VideoWriter(
    output_video_path,
    cv2.VideoWriter_fourcc(*'mp4v'),
    target_fps,
    (frame_width, frame_height)
)
previous_fires = []
frame_count = 0
with tqdm(total=estimated_processed_frames, desc="Processing frames") as pbar:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        frame_count += 1
        if frame_count % frame_interval != 0:
            continue
        try:
            processed_frame, previous_fires = process_single_frame(frame, tracker, previous_fires)
            if processed_frame is not None:
                video_writer.write(processed_frame)
            else:
                # Write blank frame on error
                blank = np.zeros((frame_height, frame_width, 3), dtype=np.uint8)
                video_writer.write(blank)
        except Exception as e:
            print(f"Error processing frame {frame_count}: {e}")
            blank = np.zeros((frame_height, frame_width, 3), dtype=np.uint8)
            video_writer.write(blank)
        pbar.update(1)
cap.release()
video_writer.release()
print(f"\nVideo saved to: {output_video_path}")
````

## 🔄 **The Quantization Process**

### **🧮 How Quantization Works:**
```
Original weight: -2.347, 1.892, 0.541, -0.238, ...
Quantized (Q4): 
  1. Find min/max range
  2. Divide into 16 levels (4 bits = 2⁴ = 16)
  3. Map each weight to nearest level
  4. Store index (0-15) instead of float32
```

### **🎨 K-Quant (K-Means Quantization):**
```python
# Traditional quantization: Equal intervals
# K-Quant: Cluster-based (smarter!)

# Step 1: Cluster weights using k-means
clusters = kmeans(weights, k=16)  # For 4-bit

# Step 2: Assign each weight to nearest cluster
# Step 3: Store cluster centers + indices
```

**🎯 Why K-Quant is better:**
- 👥 **Adapts to weight distribution**
- 🎯 **Preserves outliers better**
- 📊 **Better accuracy at same bit depth**

## 🔧 **Technical Details: GGUF Format**

### **📐 GGUF File Structure:**
```
┌─────────────────────────────────────────┐
│ 🗂️ GGUF FILE INTERNAL STRUCTURE         │
├─────────────────────────────────────────┤
│ Header: Magic number, version           │
│ Tensor Count: Number of weight tensors  │
│ Key-Value Pairs: Metadata               │
│ Tensor Data: Quantized weights          │
│ Padding: Alignment                      │
└─────────────────────────────────────────┘
```

### **🎯 Key Advantages of GGUF:**
1. **🔧 Single File**: No multiple `.bin` files
2. **📦 Efficient**: Better compression than safetensors
3. **🚀 Fast Loading**: Memory-mapped for instant startup
4. **🔒 Safe**: Validation and checksums
5. **🌐 Universal**: Works everywhere (ARM, x86, WebAssembly)

```python
Model Size × 1.2 = Minimum RAM needed

# Q4_K_M (4GB) needs ~5GB RAM
# Q8_0 (8GB) needs ~10GB RAM
# F16 (15GB) needs ~18GB RAM
```

### **3. **GPU Offloading**:**
```bash

./main -m model.gguf -ngl 35  # 35 layers on GPU

# More layers on GPU = faster but needs more VRAM
# RTX 4090 (24GB): Can offload ~40 layers of 8B model
```
